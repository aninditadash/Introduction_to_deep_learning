# __Deep Learning: Regression Problem__

```python
import pandas as pd

# split the data into train and test
from sklearn.model_selection import train_test_split

#Standardise data
from sklearn.preprocessing import StandardScaler
```

```python
# Load the data
boston_df = pd.read_csv("datasets/boston.csv")
```

```python
boston_df.head()
```

### __Exploratory Data Analysis (EDA)__

```python
from ydata_profiling import ProfileReport
```

```python
ProfileReport(boston_df)
```

```python
# Separating the Features and target variable
X = boston_df.drop(columns='HOUSEPRICE')
y = boston_df.HOUSEPRICE

# Performing split of training and test data
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)

# Performing standardisation of data
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)

X_test = scaler.transform(X_test)
```

```python
X_train
```

```python
X_train.shape
```

```python
boston_df.head()
```

```python
# Build the model
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense
from livelossplot import PlotLossesKerasTF
```

```python
# Single Layer Perceptron
model = Sequential()

# Adding the Input and Output layers
model.add(Input(shape=(X_train.shape[1],),name="Input Layer"))
model.add(Dense(1,activation='relu',name='OutputLayer'))

model.compile(loss='mse', optimizer='adam', metrics=['mae'])

# train the model
model.fit(X_train, y_train, epochs=100, validation_data=(X_test,y_test), callbacks=[PlotLossesKerasTF()])
```

```python
# Deep Neural Network (Multi Layer Perceptron)

nn = Sequential()

nn.add(Input(shape=(X_train.shape[1],), name="Input Layer"))

nn.add(Dense(15, activation='relu', name="Hidden1"))
nn.add(Dense(20, activation='relu', name="Hidden2"))
nn.add(Dense(20, activation='relu', name="Hidden3"))

nn.add(Dense(1, activation='relu', name='OutputLayer'))

# Compile the model with optimizer, loss function, and metrics
nn.compile(loss='mse', optimizer='adam', metrics=['mae'])

# train the model
nn.fit(X_train, y_train, epochs=100, validation_data=(X_test, y_test), callbacks=[PlotLossesKerasTF()])
```

### __Predicting the `X_test` data using Deep Neural n/w (Multi-layer perceptron) model__

```python
# Predicting the X_test data using Deep Neural Network

nn.predict(X_test)[:5]
```

### __Saving the model for future use__

```python
# Save the model

nn.save("BostonModel.keras")
```
