# __Artificial Neural Network (ANN)__

### __What are Neurons__

Neurons are interconnected nerve cells that build the nervous system and transmit information throughout the body. The components of a biological neural network are:

- Dendrites receive input from other neurons.
- Axons transmit the biological neuron's output.
- The cell nucleus is used for information processing.
- Synapse is the connection between two nerve cells.

### __Artificial Neuron__

It is analogous to biological neurons, where each neuron takes inputs, adds weights to them separately, sums them up, and passes this sum through a transfer function to produce a non-linear output.

<img src="https://upload.wikimedia.org/wikipedia/commons/6/60/ArtificialNeuronModel_english.png" width=500 height=500 />

$$ \text{Summation/Tranfer function} = \sum_{i=1}^{n} (W_i * X_i + b) $$

- $ W_1, W_2, \ldots, W_n $ are the weights of the connection,
- $ X_1, X_2, \ldots, X_n $ are the Inputs.
- b - Bias

* A nerve cell is a simple logic gate with binary outputs.
* Dendrites can process the input signal with a certain threshold, such as if the signal exceeds the threshold, the output signal is generated.

## __Neural Network__

Neural networks consist of interconnected computation modules that simulate the behavior of biological neurons. Each neural network consists of
multiple node layers such as:

- Input layers
- One or more hidden layers
- Output layers

ANN processes the information using the following steps:
- Inputs are passed into the first layer.
- Each individual neuron receives inputs and assigns a weight to each of them.
- The neurons generate output based on the assigned weights.
- The outputs from the first layer are subsequently forwarded to the second layer for further processing.
- The process continues until the final output is produced.


### __Types of Neural Networks__

Neural networks are used to solve complex problems that require analytical calculations. Some of the major types of neural networks are:

- Perceptron
- Multilayer perceptron
- Deep neural networks or DNNs
- Convolutional neural networks or CNNs
- Recurrent neural networks or RNNs

One can programmatically set the number and type of neural layers and the number of neurons comprising each layer. Deep neural networks consist of multiple hidden layers.

### __Perceptron:__

- This is the simplest type of ANN and is mainly used for binary prediction.
- A perceptron can only work if the data is linearly separable.

### __Multilayer perceptron (MLP):__
- It consists of multiple layers of perceptrons.
- The inputs pass through each layer, and the outputs of the last layer are the final outputs of the MLP.
- It can handle more complex problems by learning non-linear relationships.

### __Deep Neural Networks (DNN):__

A DNN is a multilayered computational model that processes data in a layered manner, refining information at each layer, analogous to how a human brain functions.

<img src="https://www.altexsoft.com/static/content-image/2024/6/ec55ad04-11ca-44f3-8eec-9399936c26ff.png" width=500 height=500 />

The depth of DNN enables it to effectively address complex problems, such as image processing and speech recognition. 

### __Convolutional Neural Network (CNN):__

CNN is a type of Deep Learning neural network architecture commonly used in Computer Vision. Computer vision is a field of Artificial Intelligence that enables a computer to understand and interpret the image or visual data. CNNs analyze visual data and are inspired by the brain's visual cortex. Effective for recognizing patterns in images.
- Example: Image Classification

### __Recurrent Neural Network (RNN):__

RNNs handle sequential data, suitable for time series and language modeling tasks. Maintain a state (memory) of previous inputs.

### __Neural Network Performance__

Deep neural network models have greater precision than conventional techniques but require more information to train and attain this precision.

### __Combination of Neural Network Layers__

Number of layers in a CNN or RNN depends on the complexity of the task and data, with deeper architectures often used for more complex problems.

- The input layer depicts the dimensions of the input vector.
- The hidden layer depicts the intermediary nodes that divide the input space into regions with soft boundaries.
- The output layer depicts the output of the neural network.
